```python
## Question 2
## Print out MainBranch types and their frequency
mainbranch_counts = {}

with open('survey_results_public.csv', 'r') as csvfile:
    reader = csv.DictReader(csvfile)  # Read the file as a list of dictionaries
    
    # Loop through each row in the CSV
    for row in reader:
        mainbranch = row['MainBranch']  # Get the 'MainBranch' value for this row
        
        # If the 'MainBranch' value already exists in the dictionary, increment its count
        if mainbranch in mainbranch_counts:
            mainbranch_counts[mainbranch] += 1
        else:
            # Initialize the count to 1 for this 'MainBranch' value
            mainbranch_counts[mainbranch] = 1

for category, count in mainbranch_counts.items():
    print(f"{category}: {count} responses")
```

```python
## Question 3
## Print out the length of your filtered list

# Find the category with the most responses
most_frequent_category = max(mainbranch_counts, key=mainbranch_counts.get)
print(f"Most frequent MainBranch category: {most_frequent_category}")

# Create a list of dictionaries containing records where 'MainBranch' matches the most frequent category
filtered_data = []

# Re-open the CSV file to filter the rows
with open('survey_results_public.csv', 'r') as csvfile:
    reader = csv.DictReader(csvfile)  # Read the file as a list of dictionaries
    
    # Loop through each row and add to the filtered data if it matches the most frequent category
    for row in reader:
        if row['MainBranch'] == most_frequent_category:
            filtered_data.append(row)

print(f"Number of records matching the most frequent 'MainBranch' category: {len(filtered_data)}")
```

```python
## Question 4
## Print out the database types and their frequency 
database_counts = {}

# Loop through the filtered data to parse and count the databases
for row in filtered_data:
    # Get the 'DatabaseHaveWorkedWith' field and split by ';' to separate individual databases
    databases = row['DatabaseHaveWorkedWith'].split(';')
    
    # Loop through each database and increment the count
    for database in databases:
        database = database.strip()  # Remove leading/trailing spaces if any
        if database:
            if database in database_counts:
                database_counts[database] += 1
            else:
                database_counts[database] = 1

for database, count in database_counts.items():
    print(f"{database}: {count} responses")
```

```python
## Question 5 
## Create new column 'DBHWW_Count', sort and print out the count of the first record
for row in filtered_data:
    # Get the 'DatabaseHaveWorkedWith' field and split by ';' to separate individual databases
    databases = row['DatabaseHaveWorkedWith'].split(';')
    
    # Count the number of databases, ignoring 'NA' responses
    database_count = sum(1 for database in databases if database.strip() != 'NA' and database.strip())
    
    # Add the count as a new column
    row['DBHWW_Count'] = database_count

# Sort the filtered data by 'DBHWW_Count' in descending order
filtered_data_sorted = sorted(filtered_data, key=lambda x: x['DBHWW_Count'], reverse=True)

print(f"DBHWW_Count of the first record: {filtered_data_sorted[0]['DBHWW_Count']}")
```

```python
## Question 6 
## Write  the first 10 rows as a TSV named '<student-first-intial><student-last-name>-module2.tsv' (insert your first and last name)
## Only include the following columns: ['ResponseId','MainBranch','DatabaseHaveWorkedWith','DBHWW_Count'] 

# Select the first 10 rows
first_10_rows = filtered_data_sorted[:10]

file_name = 'mhito-module2.tsv'

# Write the first 10 rows to a TSV file with the specified columns
with open(file_name, mode='w', newline='') as file:
    fieldnames = ['ResponseId', 'MainBranch', 'DatabaseHaveWorkedWith', 'DBHWW_Count']
    
    writer = csv.DictWriter(file, fieldnames=fieldnames, delimiter='\t')
    writer.writeheader()
    
    # Write only the desired columns for each row
    for row in first_10_rows:
        # Create a new dictionary with only the desired fields
        filtered_row = {key: row[key] for key in fieldnames}
        writer.writerow(filtered_row)

print(f"First 10 rows written to {file_name}")
```
