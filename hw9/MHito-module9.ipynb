# Flask API

## Using the POST /scrape endpoint

```python
import requests
import json

base_url = "http://127.0.0.1:5000"
topics = ["Data_engineering", "Machine_learning", "Artificial_intelligence"]

for topic in topics:
    response = requests.post(f"{base_url}/scrape", json={"topic": topic})
    print(response.content.decode())
```

## Using the DELETE /delete endpoint

```python
response = requests.delete(f"{base_url}/delete", json={"topic": "Machine_learning"})
print(response.content.decode())
```

## Using the PUT /update endpoint

```python
response = requests.put(f"{base_url}/update", json={
    "topic": "Data_engineering",
    "title": "Data Engineering",
    "content": "Data engineeringrefers to the building ofsystemsto enable the collection and usage ofdata. This data",
    "num_links": "99999"
})
print(response.content.decode())
```

## Saving Data to PostgreSQL with the POST /save endpoint

```python
response = requests.post(f"{base_url}/save")
print(response.content.decode())
```

## Retrieving Data with the GET /records endpoint

```python
response = requests.get(f"{base_url}/records")
print(response.content.decode())
```
